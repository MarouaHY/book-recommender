```python
from transformers import pipeline
import torch

# 1. Use a smaller model (faster on CPU)
model_name = "typeform/distilbert-base-uncased-mnli"  # 67MB vs BART's 1.6GB

# 2. Disable GPU explicitly and enable CPU optimizations
pipe = pipeline(
    "zero-shot-classification",
    model=model_name,
    device=torch.device("cpu"),  # Force CPU
    torch_dtype=torch.float32    # Use 32-bit precision (faster on CPU)
)
```

```python
fiction_categories = ["Fiction", "Nonfiction"]
sequence = books.loc[books["simple_categories"] == "Fiction", "description"].reset_index(drop=True)[0]
```

```python

# Run classification
result = pipe(
    sequence,
    candidate_labels=fiction_categories,
    truncation=True  # Auto-truncate long texts
)

print(result)
```

```python
import numpy as np

max_index = np.argmax(pipe(sequence, fiction_categories)["scores"])
max_label = pipe(sequence, fiction_categories)["labels"][max_index]
max_label
```

```python
def generate_predictions(sequence, categories):
    predictions = pipe(sequence, categories) # Get classification results
    max_index = np.argmax(predictions["scores"])   # Find highest probability index
    max_label = predictions["labels"][max_index]  # Return corresponding label
    return max_label
```

Classification: The function takes a text (sequence) and possible categories, then uses the Hugging Face pipeline (pipe) to classify the text. The pipeline returns a dictionary containing:

The original text ('sequence')

Your provided categories ('labels')

Probability scores ('scores') for each category (summing to 1.0)

Finding the Best Match: Using np.argmax(), it identifies which position (index) in the 'scores' list has the highest value. This index corresponds to the most likely category.

Returning the Result: The function then returns the label at that winning index position. For example, if scores are [0.7, 0.3] for ["Fiction", "Nonfiction"], it returns "Fiction" (position 0).

Takes a book description (sequence) and possible categories

Runs the classification pipeline

Returns the label with the highest score



```python
from tqdm import tqdm

actual_cats = []  # Will store true labels
predicted_cats = []  # Will store model's predictions

for i in tqdm(range(0, 300)): # Process 300 books with progress bar
        # Get i-th Fiction book description

    sequence = books.loc[books["simple_categories"] == "Fiction", "description"].reset_index(drop=True)[i]
        # Predict and store results

    predicted_cats += [generate_predictions(sequence, fiction_categories)]
    actual_cats += ["Fiction"]  # All true labels are "Fiction"
```

Initialization:

Creates two empty lists to store results

actual_cats: Will contain the known correct answers (all "Fiction")

predicted_cats: Will contain the model's guesses

The Loop:

Processes 300 books one by one (with a progress bar thanks to tqdm)

For each book:
a. Gets the book description from the DataFrame
b. Runs the classifier on it
c. Stores both the prediction and the known correct answer

Data Selection:

books.loc[books["simple_categories"] == "Fiction"] filters only fiction books

.reset_index(drop=True)[i] gets the i-th description in this filtered list

4. What You'll Get
After running this code, you'll have:

actual_cats: A list of 300 "Fiction" strings

predicted_cats: A list of 300 predictions (some "Fiction", some "Nonfiction")

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(actual_cats, predicted_cats)
print(f"Model accuracy: {accuracy:.2%}")
```

```python
isbns = []                # Will store book ISBNs
predicted_cats = []       # Will store model's predictions
```

```python
missing_cats = books.loc[books["simple_categories"].isna(), 
                        ["isbn13", "description"]].reset_index(drop=True)
#Filters rows where category is missing  then Selects only ISBN and description columns and Resets row numbering (0, 1, 2...)
```

```python
for i in tqdm(range(0, len(missing_cats))): # Processes each book with missing category:

    sequence = missing_cats["description"][i]  # Get i-th book description
    predicted_cats += [generate_predictions(sequence, fiction_categories)]  # Predict
    isbns += [missing_cats["isbn13"][i]]      # Store corresponding ISBN


```

```python
missing_predicted_df = pd.DataFrame({
    "isbn13": isbns,
    "predicted_categories": predicted_cats
})
```

```python
missing_predicted_df
```

```python
import matplotlib.pyplot as plt

# Before imputation
missing_before = books["simple_categories"].isna().sum()

# After imputation (temporary column)
books_temp = pd.merge(books, missing_predicted_df, on="isbn13", how="left")
books_temp["filled_categories"] = np.where(
    books_temp["simple_categories"].isna(), 
    books_temp["predicted_categories"], 
    books_temp["simple_categories"]
)
missing_after = books_temp["filled_categories"].isna().sum()

# Plot
fig, ax = plt.subplots(1, 2, figsize=(12, 5))
ax[0].pie([missing_before, len(books)-missing_before], 
         labels=['Missing', 'Existing'], 
         autopct='%1.1f%%',
         colors=['#ff9999','#66b3ff'])
ax[0].set_title('Categories Before Imputation')

ax[1].pie([missing_after, len(books)-missing_after], 
         labels=['Still Missing', 'Filled'], 
         autopct='%1.1f%%',
         colors=['#ff9999','#99ff99'])
ax[1].set_title('Categories After Imputation')

plt.tight_layout()
plt.show()
```

```python
# Step 1: Merge predictions back to original dataframe
books = pd.merge(
    books,                # Original dataframe
    missing_predicted_df, # Predictions dataframe  
    on="isbn13",          # Match books by ISBN
    how="left"            # Keep all original books
)

# Step 2: Fill missing categories
books["simple_categories"] = np.where(
    books["simple_categories"].isna(),  # Condition: If category is missing
    books["predicted_categories"],       # Use prediction if True
    books["simple_categories"]           # Keep original if False
)

# Step 3: Clean up
books = books.drop(columns=["predicted_categories"])
```
