```python
import numpy as np

# List of all possible emotion labels we want to track
emotion_labels = ["anger", "disgust", "fear", "joy", "sadness", "surprise", "neutral"]

# Initialize storage structures
isbn = []  # Will store book identifiers
emotion_scores = {label: [] for label in emotion_labels}  # Dictionary of empty lists for each emotion

def calculate_max_emotion_scores(predictions):
    """
    Processes raw prediction data to extract maximum scores per emotion.
    
    Args:
        predictions: List of emotion prediction results from the classifier
        
    Returns:
        Dictionary with maximum score achieved for each emotion across all predictions
    """
    # Initialize storage for all scores per emotion
    per_emotion_scores = {label: [] for label in emotion_labels}
    
    # Process each individual prediction
    for prediction in predictions:
        # Sort predictions alphabetically by emotion label
        sorted_predictions = sorted(prediction, key=lambda x: x["label"])
        
        # Store each emotion's score in its respective list
        for index, label in enumerate(emotion_labels):
            per_emotion_scores[label].append(sorted_predictions[index]["score"])
    
    # Calculate maximum score observed for each emotion
    return {label: np.max(scores) for label, scores in per_emotion_scores.items()}
```

```python
# Process the first 10 books in the dataset
for i in range(10):

    # 1. Store the ISBN-13 code of the current book
    isbn.append(books["isbn13"][i])

    # 2. Split the book's description into individual sentences
    #    (using '.' as the separator)
    sentences = books["description"][i].split(".")

    # 3. Run the classifier on all sentences
    #    This returns a list of predictions, one list of scores per sentence
    predictions = classifier(sentences)

    # 4. From all predictions, calculate the maximum score
    #    observed for each emotion across this book's sentences
    max_scores = calculate_max_emotion_scores(predictions)

    # 5. For each emotion, store its max score in emotion_scores
    #    emotion_scores[label] will hold a list of scores, one per book
    for label in emotion_labels:
        emotion_scores[label].append(max_scores[label])
```

```python
import pandas as pd
from tqdm import tqdm
from transformers import pipeline
import numpy as np

# 1. Load your book data
books = pd.read_csv("books_with_categories.csv")

# 2. Initialize the emotion classifier (CPU optimized)
classifier = pipeline(
    "text-classification",
    model="j-hartmann/emotion-english-distilroberta-base",
    device="cpu",  # Using CPU instead of MPS
    top_k=None,
    truncation=True,
    max_length=256  # Limits input size for faster processing
)

# 3. Define emotion labels and storage
emotion_labels = ["anger", "disgust", "fear", "joy", "sadness", "surprise", "neutral"]
isbn = []
emotion_scores = {label: [] for label in emotion_labels}

# 4. Function to calculate max emotion scores (same as tutorial)
def calculate_max_emotion_scores(predictions):
    per_emotion_scores = {label: [] for label in emotion_labels}
    for prediction in predictions:
        sorted_predictions = sorted(prediction, key=lambda x: x["label"])
        for index, label in enumerate(emotion_labels):
            per_emotion_scores[label].append(sorted_predictions[index]["score"])
    return {label: np.max(scores) for label, scores in per_emotion_scores.items()}

# 5. Main processing loop (optimized)
for i in tqdm(range(len(books))):
    # Store ISBN
    isbn.append(books["isbn13"][i])
    
    # Get first 3 sentences of description
    sentences = [s.strip() for s in books["description"][i].split(".") if s.strip()][:3]
    
    # Skip if no valid sentences
    if not sentences:
        for label in emotion_labels:
            emotion_scores[label].append(0.0)
        continue
    
    # Get predictions
    predictions = classifier(sentences)
    
    # Calculate max scores
    max_scores = calculate_max_emotion_scores(predictions)
    
    # Store results
    for label in emotion_labels:
        emotion_scores[label].append(max_scores[label])

# 6. Create DataFrame with results
emotions_df = pd.DataFrame(emotion_scores)
emotions_df["isbn13"] = isbn

# 7. Merge with original data
books = pd.merge(books, emotions_df, on="isbn13")

# 8. Save final results
books.to_csv("books_with_emotions.csv", index=False)
print("Emotion analysis complete!")
```
