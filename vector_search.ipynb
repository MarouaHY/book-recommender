```python
# 1. Load documents with proper encoding
from langchain.document_loaders import TextLoader
raw_documents = TextLoader("tagged_description.txt", encoding="utf-8").load()

# 2. Split text properly
from langchain.text_splitter import CharacterTextSplitter
text_splitter = CharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=200,
    separator="\n"
)
documents = text_splitter.split_documents(raw_documents)

# 3. Use local embeddings
from langchain.embeddings import HuggingFaceEmbeddings
embeddings = HuggingFaceEmbeddings(
    model_name="all-MiniLM-L6-v2",
    model_kwargs={"device": "cpu"}  # Use "cuda" if you have GPU
)

# 4. Create vector store
from langchain.vectorstores import Chroma
db_books = Chroma.from_documents(
    documents,
    embedding=embeddings,
    persist_directory="./chroma_db"  # Optional: persist to disk
)
```

```python
query = "A book to teach children about nature"
docs = db_books.similarity_search(query, k = 10)
docs
```

```python
books[books["isbn13"] == int(docs[0].page_content.split()[0].strip())]
```

```python
def retrieve_semantic_recommendations(
        query: str,
        top_k: int = 10,
) -> pd.DataFrame:
    recs = db_books.similarity_search(query, k = 50)

    books_list = []

    for i in range(0, len(recs)):
        books_list += [int(recs[i].page_content.strip('"').split()[0])]

    return books[books["isbn13"].isin(books_list)]
```

This function performs semantic search to find book recommendations based on a text query, using ISBNs from vector search results to return matching book records from your main DataFrame.

```python
retrieve_semantic_recommendations("A book to teach children about nature")
```

```python
retrieve_semantic_recommendations("A book to about love and relationship", top_k=5)
```
